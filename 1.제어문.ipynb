## 반복문 - for

> 단축키
- text 셀 -> code 셀: [esc -> y]
- code 셀 -> text 셀: [esc -> m]
- 실행: ctrl + enter
- 실행 + 다음셀로 커서 이동: shift + enter
- 셀 삭제: [esc -> dd]
- 되돌리기: z
- 현재 셀의 위쪽에 셀 추가: [esc -> a]
- 현재 셀의 아래쪽에 셀 추가: [esc -> b]


```python
# 100 이상의 숫자 출력
numbers = [273, 103, 5, 32, 65, 72, 800, 99]
over100 = []
under100 = []

for number in numbers:
    if number >= 100:
        over100.append(number)
    else:
        under100.append(number)
        

print(f"- 100 이상의 수 : {over100}")
print(f"- 100 미만의 수 : {under100}")
```

```python
numbers = input("숫자를 ','로 구분해 입력하세요:")
over100 = []
under100 = []

for number in numbers.split(","):
    if float(number) >= 100:
        over100.append(float(number))
    else:
        under100.append(float(number))


print(f"- 100 이상의 수 : {over100}")
print(f"- 100 미만의 수 : {under100}")
```

### for + dict

```python
numbers = [1,2,6,8,4,3,2,1,4,6,7,8,9,0,7,7,5,4,4,3,2,4,6,7,8,9,0,1]
counter = {}

for number in numbers:
    for i in range(10):
        if number == i:
            counter[number] = numbers.count(i)
counter = dict(sorted(counter.items()))

print(counter)
```

## list comprehensions

```python
array = [i * i for i in range(0, 20, 2)]
array
# 예측 결과: [0, 4, 16, 36, 64, 100, 144, ..., 324]
```

```python
# for문
array = []
for i in range(0, 20, 2):
    array.append(i * i)
print(array)
```

### list comprehensions + if(조건문)

```python
# 짝수만 출력
even_nums = []
for i in range(1, 20+1):
    if i % 2 == 0:
        even_nums.append(i*i)

print(even_nums)
```

```python
even_nums = [i * i for i in range(1, 20+1) if i % 2 ==0]
print(even_nums)
```

```python
key_list = ["name"]
value_list = ["기사"]

for key, value in zip(key_list, value_list):
    value_list["key"] = value
```

## 제어문 종합 문제

```python
decimal = []
for num in range(2, 101):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        decimal.append(num)


print("1~100까지의 소수 결과\n", decimal)
```
