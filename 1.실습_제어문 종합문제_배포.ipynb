```python
print("""
### 1. 고객 등급 분류 (조건문)

고객의 구매 금액에 따라 등급을 나누는 프로그램을 작성하세요.

- 100만 원 이상: **VIP**
- 50만 원 이상: **Gold**
- 10만 원 이상: **Silver**
- 그 외: **Bronze**

👉 입력: 구매 금액
👉 출력: 등급 문자열
""")
```

```python
amount = int(input("구매 금액 입력: "))


if amount >= 1000000:
    print("VIP")
elif 1000000 > amount >= 500000:
    print("Gold")
elif 500000 > amount >= 100000:
    print("Silver")
else:
    print("Bronze")
```

```python
print(
"""
### 3. 재고 소진 알림 (조건문 + 반복문)

상품 재고가 10개 있습니다.

고객이 차례로 주문할 때마다 재고를 차감하고, 재고가 0이 되면 "품절되었습니다"를 출력한 뒤 주문 처리를 중단하세요.

👉 입력: 주문 수량 리스트 `[2, 3, 5, 1]`

👉 출력: 각 주문 처리 결과 + 품절 알림
""")
```

```python
stock = 8

orders = [2, 3, 5, 1]

for order in orders:
    if order > stock:
        print("재고가 적어 주문할 수 없습니다.")
        print("품절되었습니다.")
        break

    stock -= order
    print(f"주문 처리 완료. 남은 재고는 {stock}입니다.")
    
    if stock == 0:
        print("품절되었습니다.")
        break
```
