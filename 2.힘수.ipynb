## 함수

```python
# 정의부
def mul(*values):
    result = 1
    for val in values:
        result = result * val
    print(result)

# 호출부
mul(5, 7, 9, 10)
```

```python
def comMul(num1, num2):
    for i in range(1, 101):
        if i % num1 == 0 and i % num2 ==0:
            print(f"{i}은(는) {num1}과 {num2}의 공배수입니다.")
```

```python
input_data = input("두 수를 입력하세요. 예) 2, 3")
num1, num2 = input_data.split(",")
num1, num2 = int(num1), int(num2)
comMul(num1, num2)
```
